```python
# Removing show_id and date_added columns because this is not usefull for recommodation serch
df = df.drop(['show_id','date_added'],axis=1)
```

```python
# fill na with 0 min
df['duration'] = df['duration'].fillna(df['rating'])

# Movie duration (minutes)
df['Movie_Duration'] = df['duration'].apply(
    lambda x: int(x.split()[0]) if 'min' in x else 0
)

# TV show seasons
df['Seasons'] = df['duration'].apply(
    lambda x: int(x.split()[0]) if 'Season' in x else 0
)

df = df.drop('duration',axis=1)
```

```python
df['rating']=df['rating'].apply(lambda x: np.nan if isinstance(x, str) and 'min' in x.split() else x)
```

```python
df['tags'] = (
    df['type'].fillna('') + ' ' +
    df['country'].fillna('') + ' ' +
    df['director'].fillna('') + ' ' +
    df['cast'].fillna('') + ' ' +
    df['listed_in'].fillna('') + ' ' +
    df['release_year'].fillna(0).astype(int).astype(str) + ' ' +
    df['description'].fillna('')
)

df = df[['tags','title']]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words='english')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['tags'])
vectorizer.get_feature_names_out()
print(X.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(X)
```

```python
similarity[1]
movie = "Kota Factory"

idx = df[df['title']== movie].index[0]
print(idx)
```

```python
scores = similarity[idx]
print(scores)
scores = list(enumerate(scores))
scores = sorted(scores, key=lambda x: x[1], reverse=True)
scores = scores[1:6]
print(scores)
for i in scores:
    print(df.iloc[i[0]]['title'], i[1])
```
